Esta notebook tem como objetivo resolver a competição de aprendizado de máquina "Previsão Binária de Cogumelos Venenosos" disponível no Kaggle. O problema consiste em prever se um cogumelo é comestível (c) ou venenoso (v) com base em suas características físicas.

### Objetivo do Problema
O objetivo é desenvolver um modelo de machine learning que consiga classificar corretamente os cogumelos como comestíveis ou venenosos, utilizando um conjunto de dados com características físicas dos cogumelos.

- **Coluna Alvo:** `classe` - que indica se o cogumelo é comestível ('c') ou venenoso ('v')

### Métrica de Avaliação
As submissões serão avaliadas usando o **coeficiente de correlação de Matthews (MCC)**, que é uma métrica adequada para problemas de classificação binária desbalanceados.

### Etapas do Processo
1. **Análise Exploratória dos Dados (EDA)** - compreender a estrutura e qualidade dos dados
2. **Pré-processamento dos Dados** - lidar com valores ausentes, codificação de variáveis categóricas e detecção de outliers

5. **Avaliação do Modelo** - validar o desempenho usando o conjunto de validação
6. **Previsão no Conjunto de Teste** - gerar as predições finais
7. **Criação do Arquivo de Submissão** - formatar as predições conforme o requisito do Kaggle

Este notebook seguirá as melhores práticas de machine learning, incluindo validação cruzada, tratamento adequado de dados e otimização de hiperparâmetros para obter o melhor desempenho possível.

## Tarefa 4: Identificação e Tratamento de Outliers

Nesta etapa, vamos identificar e tratar outliers nos dados de treinamento. Outliers podem distorcer os resultados do modelo e afetar negativamente o desempenho, especialmente em modelos sensíveis a valores extremos.

### Passos realizados:
1. Identificamos colunas numéricas que podem conter outliers
2. Analisamos a distribuição das colunas numéricas para detectar valores extremos
3. Utilizamos métodos estatísticos para identificar outliers (como o método IQR - Interquartile Range)
4. Avaliamos a necessidade de tratamento dos outliers

### Observações importantes:
- As colunas numéricas são: 'id', 'diametro_chapeu', 'altura_estipe', 'largura_estipe'
- Para colunas com valores extremos, podemos usar o método IQR para identificar outliers
- A decisão de tratar ou não outliers deve ser feita com cuidado, pois pode afetar o modelo
- Vamos analisar cuidadosamente os dados numéricos para entender a distribuição real

Vamos continuar com a análise de outliers nas colunas numéricas:

````python
# Identificar e tratar outliers nas colunas numéricas
numeric_columns = ['diametro_chapeu', 'altura_estipe', 'largura_estipe']

# Função para identificar outliers usando IQR
def identify_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers

# Analisar outliers para cada coluna numérica
print("Análise de outliers nas colunas numéricas:")
for column in numeric_columns:
    outliers = identify_outliers_iqr(train_df, column)
    print(f"\nColuna: {column}")
    print(f"Número de outliers: {len(outliers)}")
    print(f"Proporção de outliers: {len(outliers)/len(train_df):.4f}")
    if len(outliers) > 0:
        print("Valores extremos (outliers):")
        print(outliers[column].head())

# Para esta competição, vamos manter os dados originais e apenas verificar a presença de outliers
# Em alguns casos, outliers podem ser dados válidos, especialmente em características físicas de cogumelos
print("\nConclusão sobre outliers:")
print("- A presença de outliers é comum em dados de características físicas")
print("- Vamos manter os dados originais para preservar a informação real")
print("- O modelo deve ser robusto o suficiente para lidar com valores extremos")
````

- É crucial manter a consistência entre treinamento e teste para evitar vazamento de informação

Vamos executar a codificação das variáveis categóricas:

````python
# Identificar colunas categóricas (exceto 'classe' e 'id')
categorical_columns = [col for col in train_df.columns if train_df[col].dtype == 'object' and col not in ['classe', 'id']]
print("Colunas categóricas identificadas:")
print(categorical_columns)

# Criar cópias dos DataFrames para evitar modificações diretas
train_df_processed = train_df.copy()
test_df_processed = test_df.copy()

# Aplicar codificação one-hot nas colunas categóricas
print("\nAplicando codificação one-hot nos dados de treinamento...")
train_encoded = pd.get_dummies(train_df_processed, columns=categorical_columns, prefix=categorical_columns)

print("\nAplicando codificação one-hot nos dados de teste...")
test_encoded = pd.get_dummies(test_df_processed, columns=categorical_columns, prefix=categorical_columns)

# Garantir que ambas as codificações tenham as mesmas colunas
print("\nVerificando consistência das colunas entre treinamento e teste...")
train_columns = set(train_encoded.columns)
test_columns = set(test_encoded.columns)

# Colunas que estão em treinamento mas não em teste
missing_in_test = train_columns - test_columns
# Colunas que estão em teste mas não em treinamento
missing_in_train = test_columns - train_columns

print(f"Colunas em treinamento mas não em teste: {len(missing_in_test)}")
print(f"Colunas em teste mas não em treinamento: {len(missing_in_train)}")

# Adicionar colunas faltantes com zeros
for col in missing_in_test:
    test_encoded[col] = 0

for col in missing_in_train:
    train_encoded[col] = 0

# Reordenar colunas para garantir consistência
common_columns = sorted(list(train_columns & test_columns))
train_encoded = train_encoded[sorted(train_encoded.columns)]
test_encoded = test_encoded[sorted(test_encoded.columns)]

# Verificar o resultado
print("\nDimensões após codificação:")
print(f"Treinamento: {train_encoded.shape}")
print(f"Teste: {test_encoded.shape}")

# Verificar se a coluna 'classe' está presente no conjunto de treinamento
if 'classe' in train_encoded.columns:
    print("\nColuna 'classe' encontrada no conjunto de treinamento.")
    # Codificar a variável alvo 'classe' para 0 e 1
    train_encoded['classe'] = train_encoded['classe'].map({'c': 0, 'v': 1})
    print("Variável alvo codificada: c=0, v=1")
else:
    print("\nA coluna 'classe' não foi encontrada no conjunto de treinamento.")

print("\nCodificação concluída com sucesso!")
````

## Tarefa 9: Otimização de Hiperparâmetros do Modelo

Nesta etapa, vamos otimizar os hiperparâmetros do modelo XGBoost usando GridSearch para melhorar seu desempenho. A otimização de hiperparâmetros é crucial para alcançar o melhor desempenho possível em competições de machine learning.

### Passos realizados:
1. Preparamos os dados de treinamento e validação com codificação one-hot
2. Definimos um espaço de busca para os hiperparâmetros do XGBoost
3. Utilizamos GridSearchCV para encontrar os melhores hiperparâmetros
4. Avaliamos o desempenho do modelo otimizado usando o coeficiente de correlação de Matthews (MCC)

### Observações importantes:
- Vamos usar GridSearchCV com validação cruzada para evitar overfitting
- O espaço de busca inclui hiperparâmetros importantes como n_estimators, max_depth, learning_rate, etc.
- O objetivo é maximizar o MCC no conjunto de validação
- Como temos um grande conjunto de dados, vamos usar um subconjunto para a busca de hiperparâmetros para reduzir o tempo de execução
- O modelo otimizado será usado para previsões finais no conjunto de teste
